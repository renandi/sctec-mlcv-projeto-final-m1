### Outliers

```python
def calc_upper_outliers(df):
    # q1, q3 e IQR por coluna
    q1 = df.quantile(0.25)
    q3 = df.quantile(0.75)
    iqr = q3 - q1

    # limite superior para outliers
    upper = q3 + 1.5 * iqr

    # máscara booleana de valores > limite superior
    mask = df > upper

    # contagem de outliers por coluna
    outliers_per_col = mask.sum()

    # (opcional) linhas que possuem pelo menos um outlier
    rows_with_outlier = df[mask.any(axis=1)]

    print("Outliers superiores por coluna:")
    print(outliers_per_col)
    # print(rows_with_outlier)

    return outliers_per_col, rows_with_outlier

def calc_lower_outliers(df):
    # q1, q3 e IQR por coluna
    q1 = df.quantile(0.25)
    q3 = df.quantile(0.75)
    iqr = q3 - q1

    # limite inferior para outliers
    lower = q1 - 1.5 * iqr

    # máscara booleana de valores < limite inferior
    mask = df < lower

    outliers_per_col = mask.sum()
    rows_with_outlier = df[mask.any(axis=1)]
    print("Outliers inferiores por coluna:")
    print(outliers_per_col)

    return outliers_per_col, rows_with_outlier
```

```python
_, rows_with_upper_outliers = calc_upper_outliers(df[numerical_features])
print("--------------------------------")
_, rows_with_lower_outliers = calc_lower_outliers(df[numerical_features])
```
