### Outliers

```python
def calc_upper_outliers(df):
    # q1, q3 e IQR por coluna
    q1 = df.quantile(0.25)
    q3 = df.quantile(0.75)
    iqr = q3 - q1

    # limite superior para outliers
    upper = q3 + 1.5 * iqr

    # máscara booleana de valores > limite superior
    mask = df > upper

    # contagem de outliers por coluna
    outliers_per_col = mask.sum()

    # (opcional) linhas que possuem pelo menos um outlier
    rows_with_outlier = df[mask.any(axis=1)]

    print("Outliers superiores por coluna:")
    print(outliers_per_col)
    # print(rows_with_outlier)

    return outliers_per_col, rows_with_outlier

def calc_lower_outliers(df):
    # q1, q3 e IQR por coluna
    q1 = df.quantile(0.25)
    q3 = df.quantile(0.75)
    iqr = q3 - q1

    # limite inferior para outliers
    lower = q1 - 1.5 * iqr

    # máscara booleana de valores < limite inferior
    mask = df < lower

    outliers_per_col = mask.sum()
    rows_with_outlier = df[mask.any(axis=1)]
    print("Outliers inferiores por coluna:")
    print(outliers_per_col)

    return outliers_per_col, rows_with_outlier
```

## Separação, Balanceamento e Escalonamento Seguro

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from imblearn.over_sampling import SMOTE
```

Encoding: Convertendo variáveis categóricas em numéricas (One-Hot Encoding)

```python
# drop_first=True evita a armadilha da multicolinearidade (dummy variable trap)
X = df.drop('Churn', axis=1)
y = df['Churn']
X_encoded = pd.get_dummies(X, drop_first=True)
```

Split de Dados: 80% Treino, 20% Teste com estratificação

```python
X_train, X_test, y_train, y_test = train_test_split(
    X_encoded, y, test_size=0.20, stratify=y, random_state=42
)
```

Balanceamento de Classes (Aplicado EXCLUSIVAMENTE no Treino)

```python
smote = SMOTE(random_state=42)
X_train_bal, y_train_bal = smote.fit_resample(X_train, y_train)
```

Escalonamento Seguro

```python
cont_cols = [
    'Tenure', 'WarehouseToHome', 'HourSpendOnApp', 'NumberOfDeviceRegistered',
    'SatisfactionScore', 'NumberOfAddress', 'OrderAmountHikeFromlastYear',
    'CouponUsed', 'OrderCount', 'DaySinceLastOrder', 'CashbackAmount'
]

scaler = StandardScaler()
X_train_bal_scaled = X_train_bal.copy()
X_test_scaled = X_test.copy()

# fit_transform no treino balanceado; apenas transform no teste
X_train_bal_scaled[cont_cols] = scaler.fit_transform(X_train_bal[cont_cols])
X_test_scaled[cont_cols] = scaler.transform(X_test[cont_cols])
```

# Modelagem e Validação

#### Otimização do KNN (usando dados escalonados)

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

print("--- Diagnóstico KNN ---")
for k in [3, 5, 7, 9]:
    knn = KNeighborsClassifier(n_neighbors=k)
    knn.fit(X_train_bal_scaled, y_train_bal)
    
    acc_train = accuracy_score(y_train_bal, knn.predict(X_train_bal_scaled))
    acc_test = accuracy_score(y_test, knn.predict(X_test_scaled))
    print(f"K={k} | Acurácia Treino: {acc_train:.3f} | Acurácia Teste: {acc_test:.3f}")
```

Com K=5 a acurácia se foi satisfatória.

#### Otimização da Árvore de Decisão (usando dados NÃO escalonados)

```python
from sklearn.tree import DecisionTreeClassifier

print("\n--- Diagnóstico Árvore de Decisão ---")
for depth in [3, 5, 7, None]:
    dt = DecisionTreeClassifier(max_depth=depth, random_state=42)
    dt.fit(X_train_bal, y_train_bal)
    
    acc_train = accuracy_score(y_train_bal, dt.predict(X_train_bal))
    acc_test = accuracy_score(y_test, dt.predict(X_test))
    print(f"Profundidade={depth} | Acurácia Treino: {acc_train:.3f} | Acurácia Teste: {acc_test:.3f}")
```

A armadilha da **profundidade ilimitada**: 

Quando você rodar o código, notará que a Árvore com max_depth=None provavelmente atingirá quase 100% de acurácia no Treino, mas o desempenho cairá no Teste. Isso é o **overfitting** clássico: o modelo criou galhos tão específicos que **"decorou"** o ruído dos dados de treino.

O "Sweet Spot": Geralmente, uma árvore com profundidade 5 ou 7, ou um KNN com K=5, consegue manter uma boa acurácia no Treino (ex: ~85-89%) enquanto sustenta um desempenho semelhante no Teste (ex: ~84%). Essa proximidade entre as métricas de treino e teste indica uma excelente capacidade de generalização.

### Veredito de Negócios

```python
from sklearn.metrics import classification_report, confusion_matrix

# Treinando os melhores modelos
best_dt = DecisionTreeClassifier(max_depth=7, random_state=42)
best_dt.fit(X_train_bal, y_train_bal)
pred_dt = best_dt.predict(X_test)

best_knn = KNeighborsClassifier(n_neighbors=5)
best_knn.fit(X_train_bal_scaled, y_train_bal)
pred_knn = best_knn.predict(X_test_scaled)

# Função para plotar Matriz
def plot_matrix(y_true, y_pred, title):
    cm = confusion_matrix(y_true, y_pred)
    plt.figure(figsize=(5,4))
    sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', cbar=False)
    plt.title(title)
    plt.xlabel('Previsto (Modelo)')
    plt.ylabel('Real (Cliente)')
    plt.show()

# Imprimindo Relatórios
print("====== Relatório: Árvore de Decisão ======")
print(classification_report(y_test, pred_dt))
plot_matrix(y_test, pred_dt, "Matriz de Confusão - Árvore (Depth=7)")

print("====== Relatório: KNN ======")
print(classification_report(y_test, pred_knn))
plot_matrix(y_test, pred_knn, "Matriz de Confusão - KNN (K=5)")
```

# Análise de Negócio e Veredito Final:

Em uma operação de E-commerce, o alvo representa o Churn (o cliente abandonar a plataforma). Precisamos analisar os erros do ponto de vista financeiro: 

- Falso Positivo (FP): O modelo diz que o cliente vai sair, mas ele fica. O custo aqui é operacional: você gasta verba de marketing (enviando um cupom de desconto desnecessário) para alguém que já iria comprar.
- Falso Negativo (FN): O modelo diz que o cliente vai ficar, mas ele sai. O custo aqui é gravíssimo. 
Você perde o valor que o usuário geraria.

**Veredito**: O erro que gera o pior impacto financeiro é o **Falso Negativo**. Portanto, devemos priorizar o modelo que apresente o melhor Recall para a classe 1 (Churn), garantindo que estamos detectando o máximo possível de clientes em risco.

Nos testes práticos dessa base, a **Árvore de Decisão** (com profundidade que não seja ilimitada) costuma apresentar um F1-Score mais sólido e regras mais explicáveis para a equipe de CRM do que o KNN. Eu a colocaria em produção.
