## Visualização dos dados

### Boxplot dados numéricos

Boxplot das variáveis numéricas normalizadas para verificar presença de outliers

```python
import seaborn as sns
import matplotlib.pyplot as plt

scaled = df[numerical_features] / (df[numerical_features].max())
plt.figure(figsize=(12, 8))
sns.boxplot(data=scaled, orient='v')
plt.xticks(rotation=45, ha='right')
plt.title('Boxplots das variáveis numéricas')
plt.ylabel('Valor normalizado')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 8))
sns.boxplot(data=df[numerical_features[:-1]], orient='v')
plt.xticks(rotation=45, ha='right')
plt.title('Boxplots das variáveis numéricas')
plt.ylabel('Valor normalizado')
plt.tight_layout()
plt.show()
```

### Contagem colunas categóricas

```python
fig, axes = plt.subplots(3, 3, figsize=(18, 12))

for ax, col in zip(axes.flat, categorical_features):
    sns.countplot(data=df, x=col, ax=ax, order=df[col].value_counts().index, hue=df[col])
    ax.set_title(col)
    ax.tick_params(axis='x', rotation=45)

plt.tight_layout()
plt.show()
```

```python
fig, axes = plt.subplots(4, 3, figsize=(18, 15))

for ax, col in zip(axes.flat, numerical_features):
    sns.histplot(data=df, bins=30, ax=ax, x=col, hue=df[col])
    ax.set_title(col)
    ax.tick_params(axis='x', rotation=45)
    ax.legend_.remove()

plt.tight_layout()
plt.show()
```
